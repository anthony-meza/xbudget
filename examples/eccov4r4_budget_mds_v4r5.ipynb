```python
import numpy as np, xarray as xr
from pathlib import Path
from xmitgcm import open_mdsdataset

def open_ecco_mds_vars(var_dirs, grid_dir, prefixes=None, nx=270, nz=50,
                       ref_date="1992-01-01 12:00:00", iters="all",
                       dtype=np.dtype(">f4"), chunks=None, compute_grid=False):
    grid_dir = Path(grid_dir)
    var_dirs = [Path(p) for p in np.atleast_1d(var_dirs)]
    prefixes = [p.name for p in var_dirs] if prefixes is None else prefixes
    prefixes = [prefixes] if isinstance(prefixes, str) else prefixes
    if len(prefixes) != len(var_dirs):
        raise ValueError("prefixes must have same length as var_dirs")

    chunks = chunks or {d: -1 for d in (
        "tile", "face", "k", "k_l", "k_u", "k_p1", "i", "i_g", "j", "j_g"
    )} | {"time": 1}

    dsets = []
    for var_dir, prefix in zip(var_dirs, prefixes):
        ds_i = open_mdsdataset(
            data_dir=str(var_dir), grid_dir=str(grid_dir), prefix=[prefix],
            geometry="llc", nx=nx, nz=nz, iters=iters, read_grid=False,
            swap_dims=False, ref_date=ref_date, default_dtype=dtype,
            chunks={d: c for d, c in chunks.items() if d != "tile"},
        )
        if "face" in ds_i.dims:
            ds_i = ds_i.rename({"face": "tile"})
        dsets.append(ds_i.chunk({d: c for d, c in chunks.items() if d in ds_i.dims}))

    ds = xr.merge(dsets, compat="override", join="exact")

    ds_grid = xr.open_dataset(
        grid_dir / "ECCO-GRID.nc",
        chunks={d: c for d, c in chunks.items() if d != "time"},
    )
    if compute_grid:
        ds_grid = ds_grid.compute()
    if "face" in ds_grid.dims and "tile" not in ds_grid.dims:
        ds_grid = ds_grid.rename({"face": "tile"})
    if ds_grid.sizes.get("tile") != ds.sizes["tile"]:
        raise ValueError(f"tile mismatch: ds={ds.sizes['tile']}, grid={ds_grid.sizes.get('tile')}")

    ds_grid = ds_grid.assign_coords(tile=ds.tile)
    return xr.merge([ds, ds_grid], compat="override", join="exact").chunk(
        {d: c for d, c in chunks.items() if d in ds.dims or d in ds_grid.dims}
    )
```

```python
def get_trsp(ds, vel):
    if "i_g" in vel.dims:
        trsp = vel.fillna(0.0) * ds["drF"] * ds["dyG"]
        return trsp.where(ds["maskW"]).fillna(0.0)

    if "j_g" in vel.dims:
        trsp = vel.fillna(0.0) * ds["drF"] * ds["dxG"]
        return trsp.where(ds["maskS"]).fillna(0.0)

    if "k_l" in vel.dims:
        trsp = vel.fillna(0.0) * ds["rA"]
        return trsp.fillna(0.0)

    raise ValueError("Expected velocity on `i_g`, `j_g`, or `k_l`.")
```

```python
def plot_interpolated_ecco(fig, ax, ds, vmin=0, vmax=30, cmap="coolwarm", target_lon= None, target_lat = None, method = "nearest"):
    # Build a simple regular lat-lon target grid for quick visual comparison.
    if target_lon is None:
        target_lon = np.arange(-180, 180, 1)
    if target_lat is None: 
        target_lat = np.arange(-90, 90, 1)
    lon_grid, lat_grid = np.meshgrid(target_lon, target_lat)
    
    # Flatten the LLC field into point/value pairs for interpolation.
    points = np.column_stack((ds.XC.values.ravel(), ds.YC.values.ravel()))
    values = ds.values.ravel()
    
    # Use nearest-neighbor remapping to preserve the native diagnostic values.
    interpolated_data = griddata(points, values, (lon_grid, lat_grid), method=method)
    
    # Plot the remapped field.
    cb = ax.pcolormesh(lon_grid, lat_grid, interpolated_data, 
                       vmin=vmin, vmax=vmax, cmap=cmap)
    
    return cb
ds["mask"] = mask_k.where(mask_k > 0.0).fillna(0.0)
fig, ax = plt.subplots(figsize=(12, 6))
# plot_interpolated_ecco(fig, ax, ds["mask"])
plot_interpolated_ecco(fig, ax, (ds["Depth"] < 1000) * (ds["Depth"] > 0.0))
```

```python
def reverse_cumsum(da, dim):
    return da.isel({dim: slice(None, None, -1)}).cumsum(dim).isel({dim: slice(None, None, -1)})
```

```python
psi = reverse_cumsum(UVRES_CONV.sum(["tile", "j", "i"]), "SIG2")
psi = psi.compute()
```

```python
fig, ax = plt.subplots(figsize=(5, 5))

psi_sel = 1e-6 * psi.sel(SIG2=slice(33, 39))
psi_sel = psi_sel.groupby("time.year").mean("time")

q_lo = psi_sel.quantile(0.025, dim="year")
q_med = psi_sel.quantile(0.5, dim="year")
q_hi = psi_sel.quantile(0.975, dim="year")

line = q_med.plot(y="SIG2", ax=ax)[0]

patch = ax.fill_betweenx(
    q_med["SIG2"],
    q_lo,
    q_hi,
    alpha=0.25,
)

ax.set_title("median cross-ACS overturning (quantile spread 2.5-97.5%)")
ax.invert_yaxis()
ax.set_xlabel("Overturning [Sv]")
ax.set_ylabel(r"$\sigma_2$ [kg m$^{-3}$]")
ax.grid(alpha=0.2)
```
